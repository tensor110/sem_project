# 01 — Setup
Mount Drive, check GPU, install dependencies, download dataset.

## 1. Mount Google Drive

```python
from google.colab import drive
drive.mount('/content/drive')
```

## 2. Check GPU

```python
import torch
print('GPU available :', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU name      :', torch.cuda.get_device_name(0))
```

## 3. Create project folders in Drive

```python
import os

BASE = '/content/drive/MyDrive/ContentRecognition'
folders = [
    f'{BASE}/checkpoints/image',
    f'{BASE}/results/image',
]
for f in folders:
    os.makedirs(f, exist_ok=True)

print('Folders ready:')
for f in folders:
    print(' ', f)
```

## 4. Upload Kaggle API key and configure

```python
from google.colab import files

# Upload your kaggle.json when the file picker appears
files.upload()

os.makedirs('/root/.config/kaggle', exist_ok=True)
os.system('cp kaggle.json /root/.config/kaggle/')
os.system('chmod 600 /root/.config/kaggle/kaggle.json')
print('Kaggle configured!')
```

## 5. Download Intel Image Classification dataset

```python
# Download to /content (fast local SSD)
os.system('kaggle datasets download -d puneet6060/intel-image-classification -p /content')
os.system('unzip -q /content/intel-image-classification.zip -d /content/intel_data')
print('Dataset extracted!')
```

## 6. Verify folder structure

```python
data_root = '/content/intel_data'
for root, dirs, files_list in os.walk(data_root):
    level = root.replace(data_root, '').count(os.sep)
    if level < 3:
        indent = '  ' * level
        print(f'{indent}{os.path.basename(root)}/')
        if level == 2:
            print(f'{indent}  → {len(files_list)} images')
```
